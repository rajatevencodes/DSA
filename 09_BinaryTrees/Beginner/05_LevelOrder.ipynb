```python
"""LevelOrder"""
from queue import Queue
class Node:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None


class Solution:
    def levelOrder(self, root):
        ans = []
        q = Queue()
        q.put(root)

        while not q.empty():
            level = []
            size = q.qsize()

            for _ in range(size):
                node = q.get()
                level.append(node.val)

                if node.left:
                    q.put(node.left)
                if node.right:
                    q.put(node.right)

            ans.append(level)

        return ans


if __name__ == "__main__":
    root = Node(1)
    root.left = Node(2)
    root.right = Node(3)
    root.left.left = Node(4)
    root.left.right = Node(5)
    solution = Solution()
    print("Level-Order Traversal:")
    print(solution.levelOrder(root))
```
