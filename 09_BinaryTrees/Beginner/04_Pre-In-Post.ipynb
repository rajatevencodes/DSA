```python
"""PreOrder-InOrder-PostOrder"""

class Node:
    def __init__(self, val):
        self.val = val
        self.left = None
        self.right = None


class Solution:
    def tree_traversal(self, root):
        pre, in_order, post = [], [], []
        stack = [(root, 1)]

        while stack:
            node, state = stack.pop()
            if state == 1:
                pre.append(node.val)
                stack.append((node, 2))
                if node.left:
                    stack.append((node.left, 1))
            elif state == 2:
                in_order.append(node.val)
                stack.append((node, 3))
                if node.right:
                    stack.append((node.right, 1))
            else:
                post.append(node.val)

        return [pre, in_order, post]


if __name__ == "__main__":
    root = Node(1)
    root.left = Node(2)
    root.right = Node(3)
    root.left.left = Node(4)
    root.left.right = Node(5)
    root.right.left = Node(6)
    root.right.right = Node(7)

    sol = Solution()
    pre, in_order, post = sol.tree_traversal(root)
    print(pre)
    print(in_order)
    print(post)
```
