```python
from queue import Queue


class Node:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


class Solution:
    def zigzagLevelOrder(self, root):
        ans = []
        q = Queue()
        q.put(root)
        left_to_right = True  # toggle this each level

        while not q.empty():
            level = []
            size = q.qsize()
            for _ in range(size):
                node = q.get()
                level.append(node.val)
                if node.left:
                    q.put(node.left)
                if node.right:
                    q.put(node.right)

            # if current level should be right->left, reverse the list
            if not left_to_right:
                level.reverse()

            ans.append(level)
            left_to_right = not left_to_right  # toggle direction

        return ans


if __name__ == "__main__":
    root = Node(1, left=Node(2, Node(4), Node(5)), right=Node(3))
    print(Solution().zigzagLevelOrder(root))
```
