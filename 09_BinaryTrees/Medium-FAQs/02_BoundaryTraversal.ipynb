```python
class Node:
    def __init__(self, data=0, left=None, right=None):
        self.data = data
        self.left = left
        self.right = right


class Solution:
    def boundary(self, root):
        if not root:
            return []

        self.result_nodes = []

        # If the root is a leaf, it will be picked up by the leaf traversal.
        if not self.is_leaf(root):
            self.result_nodes.append(root.data)

        # Step 1: Add all nodes of the left boundary.
        self.add_left_boundary(root)

        # Step 2: Add all leaf nodes from left to right.
        self.add_leaf_nodes(root)

        # Step 3: Add all nodes of the right boundary in reverse order.
        self.add_right_boundary(root)

        return self.result_nodes

    def add_left_boundary(self, root):
        # Start from the root's left child.
        current_node = root.left
        while current_node:
            # Add the node's value only if it's not a leaf.
            if not self.is_leaf(current_node):
                self.result_nodes.append(current_node.data)

            # Prioritize moving left. If there's no left child, move right.
            if current_node.left:
                current_node = current_node.left
            else:
                current_node = current_node.right

    def add_leaf_nodes(self, root):
        # Base case: if the node is null, stop.
        if not root:
            return

        # If the current node is a leaf, add its value to our result.
        if self.is_leaf(root):
            self.result_nodes.append(root.data)

        # Recursively find leaves in the left and right subtrees
        self.add_leaf_nodes(root.left)
        self.add_leaf_nodes(root.right)

    def add_right_boundary(self, root):
        current_node = root.right

        # This list temporarily holds the right boundary nodes in top-down order.
        temp_right_nodes = []
        while current_node:
            # Add the node's value only if it's not a leaf.
            if not self.is_leaf(current_node):
                temp_right_nodes.append(current_node.data)

            # Prioritize moving right. If there's no right child, move left.
            if current_node.right:
                current_node = current_node.right
            else:
                current_node = current_node.left

        # Add the collected nodes in reverse order (bottom-up).
        self.result_nodes.extend(temp_right_nodes[::-1])

    def is_leaf(self, node):
        # A node is a leaf if it has no left and no right children.
        if node.left is None and node.right is None:
            return True
        return False


if __name__ == "__main__":
    root = Node(1, left=Node(2, Node(4), Node(5)), right=Node(3))
    print(Solution().boundary(root))
```
