```python
class Node:
    def __init__(self, val=0, left=None, right=None):
        self.data = val
        self.left = left
        self.right = right


class Solution:
    def is_symmetric(self, root):
        if not root:
            return True  # An empty tree is symmetric
        return self.symmetry(root.left, root.right)

    def symmetry(self, left, right):
        # Both nodes are null, so symmetric
        if not left and not right:
            return True
        # One of the nodes is null, so not symmetric
        if not left or not right:
            return False

        # The values of the nodes do not match, so not symmetric
        if left.data != right.data:
            return False

        return self.symmetry(left.left, right.right) and self.symmetry(left.right, right.left)

if __name__ == "__main__":
    root = Node(1, left=Node(2, Node(4), Node(5)), right=Node(3))
    root2 = Node(1, left=Node(2),right=Node(2))
    sol = Solution()
    print(sol.is_symmetric(root))
    print(sol.is_symmetric(root2))
```
