```python
"""PreRequisite"""
class Node:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


class Solution:
    def result(self, root):
        if root is None:
            return 0
        left_sum = self.result(root.left)
        right_sum = self.result(root.right)
        return root.val + left_sum + right_sum


# Example tree:
#       1
#      / \
#     2   3
#    / \
#   4   5
if __name__ == "__main__":
    root = Node(1, left=Node(2, Node(4), Node(5)), right=Node(3))
    sol = Solution()
    print(sol.result(root))  
```

```python
class Node:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

class Solution:
    def maxPathSum(self, root):
        self.currentMaxSum = float("-inf")
        self.result(root)
        return self.currentMaxSum

    def result(self, root):
        if root is None:
            return 0
        left = max(self.result(root.left), 0)
        right = max(self.result(root.right), 0)
        self.currentMaxSum = max(self.currentMaxSum, root.val + left + right)
        return root.val + max(left, right)

# Example tree:
#       1
#      / \
#     2   3
#    / \
#   4   5
if __name__ == "__main__":
    root = Node(1, left=Node(2, Node(4), Node(5)), right=Node(3))
    sol = Solution()
    print(sol.maxPathSum(root))
```
