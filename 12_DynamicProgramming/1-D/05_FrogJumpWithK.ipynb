```python
"""Recurssion"""
class Solution:
    def frogJump(self, heights, k):
        index = len(heights) - 1
        return self.result(heights, k, index)

    def result(self, heights, k, index):
        # Base call
        if index == 0:
            return 0

        # Recurssive call + Kaam
        minimumEnergy = float("inf")

        for j in range(1, k + 1):
            if (index - j) >= 0:  # Non-negative only
                jumpKEnergy = self.result(heights, k, index - j) + abs(
                    heights[index] - heights[index - j]
                )
                minimumEnergy = min(minimumEnergy, jumpKEnergy)

        return minimumEnergy


if __name__ == "__main__":
    heights = [15, 4, 1, 14, 15]
    k = 3
    sol = Solution()
    print("Minimum energy:", sol.frogJump(heights, k))
```

```python
"""Memoization"""

class Solution:
    def frogJump(self, heights, k):
        n = len(heights)
        dp = [None] * n
        dp[0] = 0
        return self.result(heights, k, n - 1, dp)

    def result(self, heights, k, index, dp):
        # Base Call
        if index == 0:
            return 0

        if dp[index] != None:
            return dp[index]

        # Recurssive Call + Kaam
        minimumEnergy = float("inf")

        for j in range(1, k + 1):
            if (index - j) >= 0:
                jumpKEnergy = self.result(heights, k, index - j, dp) + abs(
                    heights[index] - heights[index - j]
                )
                minimumEnergy = min(minimumEnergy, jumpKEnergy)

        dp[index] = minimumEnergy
        return dp[index]

if __name__ == "__main__":
    heights = [15, 4, 1, 14, 15]
    k = 3
    sol = Solution()
    print("Minimum energy:", sol.frogJump(heights, k))
```

```python
"""Tabulation"""
class Solution:
    def frogJump(self, heights, k):
        n = len(heights)
        dp = [None] * n
        dp[0] = 0

        for i in range(1, n):
            minimumEnergy = float("inf")
            for j in range(1, k + 1):
                if (i - j) >= 0:
                    jumpKEnergy = dp[i - j] + abs(heights[i] - heights[i - j])
                    minimumEnergy = min(minimumEnergy, jumpKEnergy)
            dp[i] = minimumEnergy

        return dp[n - 1]

if __name__ == "__main__":
    heights = [15, 4, 1, 14, 15]
    k = 3
    sol = Solution()
    print("Minimum energy:", sol.frogJump(heights, k))
```
