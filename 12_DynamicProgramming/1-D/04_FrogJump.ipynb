```python
class Solution:
    def frogJump(self, heights):
        index = len(heights) - 1
        return self.result(heights, index)

    def result(self, heights, index):
        # Base call
        if index == 0:
            return 0
        # Recurssive Call + Kaam
        jumpOne = self.result(heights, index - 1) + abs(
            heights[index] - heights[index - 1]
        )
        jumpTwo = float("inf")  # To make access local scope variable in parent scope
        if index > 1:
            jumpTwo = self.result(heights, index - 2) + abs(
                heights[index] - heights[index - 2]
            )

        return min(jumpOne, jumpTwo)

if __name__ == "__main__":
    heights = [10, 20, 30, 20]
    obj = Solution()
    print(obj.frogJump(heights))  # Output: 10
```

```python
"""Memoization"""
class Solution:
    def frogJump(self, heights):
        index = len(heights) - 1
        dp = [None] * (len(heights) + 1)
        dp[0] = 0
        return self.result(heights, index, dp)

    def result(self, heights, index, dp):
        # Base call
        if index == 0:
            return 0
        if dp[index] != None:
            return dp[index]
        # Recurssive Call + Kaam
        jumpOne = self.result(heights, index - 1, dp) + abs(
            heights[index] - heights[index - 1]
        )
        jumpTwo = float("inf")
        if index > 1:
            jumpTwo = self.result(heights, index - 2, dp) + abs(
                heights[index] - heights[index - 2]
            )
        dp[index] = min(jumpOne, jumpTwo)
        return dp[index]


if __name__ == "__main__":
    heights = [10, 20, 30, 20]
    obj = Solution()
    print(obj.frogJump(heights))  # Output: 10
```

```python
class Solution:
    def frogJump(self, heights):
        dp = [None] * len(heights)
        dp[0] = 0
        for i in range(1, len(heights)):
            jumpOne = dp[i - 1] + abs(heights[i] - heights[i - 1])
            jumpTwo = float("inf")
            if i > 1:
                jumpTwo = dp[i - 2] + abs(heights[i] - heights[i - 2])
            dp[i] = min(jumpOne, jumpTwo)
        return dp[len(heights) - 1]

if __name__ == "__main__":
    heights = [10, 20, 30, 20]
    obj = Solution()
    print(obj.frogJump(heights))  # Output: 10
```
