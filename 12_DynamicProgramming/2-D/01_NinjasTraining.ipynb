```python
"""Recurssion"""
class Solution:
    def ninjaTraining(self, matrix):
        # Start the recursion from Day 0.
        # '3' is a placeholder for the previous day's activity, as there was none.
        return self.solve(matrix, 0, 3)

    def solve(self, matrix, day, last_activity):
        # Base Case: If we are past the last day, no more points can be added.
        if day == len(matrix):
            return 0

        max_points = 0
        # For the current day, try all possible activities.
        for activity in range(3):
            # Check if the current activity is different from the last one.
            if activity != last_activity:
                # Calculate points for this path:
                # current activity's points + points from all future days.
                points = matrix[day][activity] + self.solve(matrix, day + 1, activity)

                # Update the maximum points found.
                max_points = max(max_points, points)

        return max_points
```

```python
"""Memoization"""
class Solution:
    def ninjaTraining(self, matrix):
        n = len(matrix)
        # DP table to store results. Initialize with -1 (uncalculated).
        # Size is n x 4 to handle days 0 to n-1 and last_activity 0, 1, 2, and 3 (initial state).
        dp = [[-1] * 4 for _ in range(n)]

        # Start the recursion from Day 0. '3' is a placeholder for the initial state.
        return self.solve(matrix, 0, 3, dp)

    def solve(self, matrix, day, last_activity, dp):
        # Base Case: Past the last day.
        if day == len(matrix):
            return 0

        # Step 1: If this state is already calculated, return the stored value.
        if dp[day][last_activity] != -1:
            return dp[day][last_activity]

        max_points = 0
        for activity in range(3):
            if activity != last_activity:
                points = matrix[day][activity] + self.solve(
                    matrix, day + 1, activity, dp
                )
                max_points = max(max_points, points)

        # Step 2: Store the result in the DP table before returning.
        dp[day][last_activity] = max_points
        return max_points
```

```python
"""Tabulation"""
class Solution:
    def ninjaTraining(self, matrix):
        n = len(matrix)
        # DP table of size (n+1) x 4. The extra row handles the base case.
        dp = [[0] * 4 for _ in range(n + 1)]

        # Iterate from the last day backwards to the first day.
        for day in range(n - 1, -1, -1):
            for last_activity in range(4):
                max_points = 0
                for activity in range(3):
                    if activity != last_activity:
                        # Get points from the pre-calculated next day's results.
                        points = matrix[day][activity] + dp[day + 1][activity]
                        max_points = max(max_points, points)
                dp[day][last_activity] = max_points

        # The answer is the result for day 0 with no prior activity constraint.
        return dp[0][3]
```
