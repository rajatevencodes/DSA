```python
"""Recurssion"""
class Solution:
    def minFallingPathSum(self, matrix):
        maxRows = len(matrix)
        maxCols = len(matrix[0])
        minPath = float("inf")

        for col in range(maxCols):
            pathSum = self.result(maxRows - 1, col, maxCols, matrix)
            minPath = min(minPath, pathSum)

        return minPath

    def result(self, row, col, maxCols, matrix):
        if col < 0 or col >= maxCols:
            return int(1e9)
        if row == 0:
            return matrix[row][col]
        up = matrix[row][col] + self.result(row - 1, col, maxCols, matrix)
        left_diag = matrix[row][col] + self.result(row - 1, col - 1, maxCols, matrix)
        right_diag = matrix[row][col] + self.result(row - 1, col + 1, maxCols, matrix)
        return min(up, left_diag, right_diag)


if __name__ == "__main__":
    matrix = [[1, 2, 10, 4], [100, 3, 2, 1], [1, 1, 20, 2], [1, 2, 2, 1]]
    sol = Solution()
    print(sol.minFallingPathSum(matrix))
```

```python
"""Memorization"""
class Solution:
    def minFallingPathSum(self, matrix):
        maxRows = len(matrix)
        maxCols = len(matrix[0])
        minPath = float("inf")
        dp = [[-1 for _ in range(maxCols)] for _ in range(maxRows)]

        for col in range(maxCols):
            pathSum = self.result(maxRows - 1, col, maxCols, matrix, dp)
            minPath = min(minPath, pathSum)

        return minPath

    def result(self, row, col, maxCols, matrix, dp):
        if col < 0 or col >= maxCols:
            return int(1e9)
        if row == 0:
            return matrix[row][col]
        if dp[row][col] != -1:
            return dp[row][col]
        up = matrix[row][col] + self.result(row - 1, col, maxCols, matrix, dp)
        left_diag = matrix[row][col] + self.result(row - 1, col - 1, maxCols, matrix, dp)
        right_diag = matrix[row][col] + self.result(row - 1, col + 1, maxCols, matrix, dp)
        dp[row][col] = min(up, left_diag, right_diag)
        return dp[row][col]


if __name__ == "__main__":
    matrix = [[1, 2, 10, 4], [100, 3, 2, 1], [1, 1, 20, 2], [1, 2, 2, 1]]
    sol = Solution()
    print(sol.minFallingPathSum(matrix))
```
