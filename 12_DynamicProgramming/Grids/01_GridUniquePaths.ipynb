```python
"""
Recurssion Approach
    (0,0) to (m-1,n-1) - Bottom up
    (m-1,n-1) to (0,0) - Top down
"""

class Solution:
    def uniquePaths(self, m: int, n: int) -> int:
        return self.result(m-1 , n-1,"")

    def result(self,currentRow, currentCol,ans):
        # Base Case 1: If we reach the destination, we've found 1 valid path.
        if currentRow == 0 and currentCol == 0:
            print(ans + "End")
            return 1

        # Base Case 2: If we go out of bounds, this path is invalid (0 paths found).
        if currentRow < 0 or currentCol < 0:
            return 0

        # Recursive Step: The total paths from the current cell are the sum of
        # paths from moving down and paths from moving right.
        paths_from_down = self.result(currentRow - 1, currentCol, ans + "Down->")
        paths_from_right = self.result(currentRow, currentCol - 1, ans + "Right->")

        return paths_from_down + paths_from_right

if __name__ == "__main__":
    m = 3
    n = 2
    sol = Solution()
    print("Number of ways:", sol.uniquePaths(m, n))
```
