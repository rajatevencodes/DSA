```python
"""Recurssion"""
class Solution:
    def uniquePathsWithObstacles(self, matrix):
        maxRow = len(matrix)
        maxCol = len(matrix[0])
        return self.result(maxRow - 1, maxCol - 1, matrix)

    def result(self, currentRow, currentCol, matrix):
        if matrix[currentRow][currentCol] == 1:
            return 0
        if currentRow == 0 and currentCol == 0:
            return 1
        if currentRow < 0 or currentCol < 0:
            return 0
        up = self.result(currentRow - 1, currentCol, matrix)
        left = self.result(currentRow, currentCol - 1, matrix)
        return up + left

if __name__ == "__main__":
    maze = [[0, 0, 0], [0, 1, 0], [0, 0, 0]]
    sol = Solution()
    print("Number of paths with obstacles:", sol.uniquePathsWithObstacles(maze))
```

```python
"""Memorization"""
class Solution:
    def uniquePathsWithObstacles(self, matrix):
        maxRow = len(matrix)
        maxCol = len(matrix[0])
        dp = [[-1 for j in range(maxCol)] for i in range(maxRow)]
        return self.result(maxRow - 1, maxCol - 1, matrix, dp)

    def result(self, currentRow, currentCol, matrix, dp):
        if matrix[currentRow][currentCol] == 1:
            return 0
        if currentRow == 0 and currentCol == 0:
            return 1
        if currentRow < 0 or currentCol < 0:
            return 0
        if dp[currentRow][currentCol] != -1:
            return dp[currentRow][currentCol]

        up = self.result(currentRow - 1, currentCol, matrix, dp)
        left = self.result(currentRow, currentCol - 1, matrix, dp)
        dp[currentRow][currentCol] = up + left
        return dp[currentRow][currentCol]

if __name__ == "__main__":
    maze = [[0, 0, 0], [0, 1, 0], [0, 0, 0]]
    sol = Solution()
    print("Number of paths with obstacles:", sol.uniquePathsWithObstacles(maze))
```
