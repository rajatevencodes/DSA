```python
"""Recurssion"""
class Solution:
    def minTriangleSum(self, triangle):
        n = len(triangle)
        return self.result(0, 0, triangle, n)

    def result (self, i, j, triangle, n):
        if i == n - 1:
            return triangle[i][j]
        down = triangle[i][j] + self.result (i + 1, j, triangle, n)
        diagonal = triangle[i][j] + self.result (i + 1, j + 1, triangle, n)
        return min(down, diagonal)

if __name__ == "__main__":
    triangle = [[1], [2, 3], [3, 6, 7], [8, 9, 6, 10]]
    sol = Solution()
    print("Minimum path sum in triangle:", sol.minTriangleSum(triangle))
```

```python
"""Memorization"""
class Solution:
    def minTriangleSum(self, triangle):
        n = len(triangle)
        dp = [[-1 for _ in range(n)] for _ in range(n)]
        return self.result(0, 0, triangle, n, dp)

    def result(self, i, j, triangle, n, dp):
        if i == n - 1:
            return triangle[i][j]
        if dp[i][j] != -1:
            return dp[i][j]
        down = triangle[i][j] + self.result(i + 1, j, triangle, n, dp)
        diagonal = triangle[i][j] + self.result(i + 1, j + 1, triangle, n, dp)
        dp[i][j] = min(down, diagonal)
        return dp[i][j]

if __name__ == "__main__":
    triangle = [[1], [2, 3], [3, 6, 7], [8, 9, 6, 10]]
    sol = Solution()
    print("Minimum path sum in triangle:", sol.minTriangleSum(triangle))
```
