```python
class Solution:
    def stockBuySell(self, arr, n, fee):
        canWeBuy = True
        return self.result(0, canWeBuy, n, arr, fee)

    def result(self, ind, canWeBuy, n, arr, fee):
        if ind == n:
            return 0

        profit = 0

        if canWeBuy == True:
            pick = -arr[ind] + self.result(ind + 1, False, n, arr, fee)
            notPick = 0 + self.result(ind + 1, True, n, arr, fee)
            profit = max(pick, notPick)

        if canWeBuy == False:
            pick = arr[ind] - fee + self.result(ind + 1, True, n, arr, fee)
            notPick = 0 + self.result(ind + 1, False, n, arr, fee)
            profit = max(pick, notPick)

        return profit


if __name__ == "__main__":
    n = 6
    arr = [10, 3, 7, 5, 1, 3]
    fee = 3
    sol = Solution()
    print("The maximum profit that can be generated is", sol.stockBuySell(arr, n, fee))
```
