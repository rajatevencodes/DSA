```python
"""Recurssion"""
class Solution:
    def stockBuySell(self, arr, n):
        canWeBuy = True
        return self.result(0, canWeBuy, n, arr)

    # Function to find the maximum profit earned using recursion
    def result(self, ind, canWeBuy, n, arr):
        # Base case: If we are past the last day, no more profit can be made.
        if ind == n:
            return 0

        profit = 0

        if canWeBuy == True:
            # We have two choices:
            # 1. Buy the stock today.
            # (profit = sell - buy) where buy will always be negative
            pick = -arr[ind] + self.result(ind + 1, False, n, arr)
            # 2. Don't buy today.
            notPick = 0 + self.result(ind + 1, True, n, arr)
            profit = max(pick, notPick)

        if canWeBuy == False:
            # We have two choices:
            # 1. Sell the stock today (pick).
            pick = arr[ind] + self.result(ind + 1, True, n, arr)
            # 2. Don't sell today (notPick).
            notPick = 0 + self.result(ind + 1, False, n, arr)
            profit = max(pick, notPick)

        # Return the maximum calculated profit for the current state
        return profit

if __name__ == "__main__":
    n = 6
    arr = [7, 1, 5, 3, 6, 4]
    sol = Solution()
    print("The maximum profit that can be generated is", sol.stockBuySell(arr, n))
```

```python
"""Memoization"""
# TODO : dp part
class Solution:
    def stockBuySell(self, arr, n):
        canWeBuy = True
        # 0 - can buy 1 - can sell
        dp = [[-1 for _ in range(2)] for _ in range(n)]
        return self.result(0, canWeBuy, n, arr, dp)

    def result(self, ind, canWeBuy, n, arr, dp):
        if ind == n:
            return 0

        if dp[ind][int(canWeBuy)] != -1:
            return dp[ind][int(canWeBuy)]

        profit = 0

        if canWeBuy == True:
            pick = -arr[ind] + self.result(ind + 1, False, n, arr, dp)
            notPick = 0 + self.result(ind + 1, True, n, arr, dp)
            profit = max(pick, notPick)

        if canWeBuy == False:
            pick = arr[ind] + self.result(ind + 1, True, n, arr, dp)
            notPick = 0 + self.result(ind + 1, False, n, arr, dp)
            profit = max(pick, notPick)

        dp[ind][int(canWeBuy)] = profit
        return profit

if __name__ == "__main__":
    n = 5
    arr = [8, 6, 5, 4, 3]
    sol = Solution()
    print("The maximum profit that can be generated is", sol.stockBuySell(arr, n))
```
