```python
"""Recurssion"""
class Solution:
    def stockBuySell(self, arr, n):
        canWeBuy = True
        return self.result(0, canWeBuy, n, arr, 2)

    def result(self, ind, canWeBuy, n, arr, cap):
        if ind == n or cap == 0:
            return 0

        profit = 0

        if canWeBuy == True:
            pick = -arr[ind] + self.result(ind + 1, False, n, arr, cap)
            notPick = 0 + self.result(ind + 1, True, n, arr, cap)
            profit = max(pick, notPick)

        if canWeBuy == False:
            pick = arr[ind] + self.result(ind + 1, True, n, arr, cap - 1)
            notPick = 0 + self.result(ind + 1, False, n, arr, cap)
            profit = max(pick, notPick)

        # Return the maximum calculated profit for the current state
        return profit

if __name__ == "__main__":
    n = 6
    arr = [5, 7, 2, 10, 6, 9]
    sol = Solution()
    print("The maximum profit that can be generated is", sol.stockBuySell(arr, n))
```
